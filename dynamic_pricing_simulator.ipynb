```python
import pandas as pd
import numpy as np

# 1. Define our product catalog data with costs, current prices, competitor prices, and stock levels
data = {
    'product_id': ["PROD_101", "PROD_102", "PROD_103", "PROD_104", "PROD_105"],
    'category': ["Electronics", "Home & Living", "Electronics", "Apparel", "Home & Living"],
    'our_cost': [80.00, 50.00, 150.00, 25.00, 90.00],        # What the item costs us wholesale
    'our_price': [120.50, 85.00, 210.00, 45.99, 150.00],     # Current selling price
    'competitor_price': [130.00, 80.00, 205.00, 50.00, 140.00], # Competitor price
    'stock_level': [15, 120, 8, 250, 45]                     # Inventory count
}

# Convert our dictionary into a structured Pandas DataFrame
df_pricing = pd.DataFrame(data)

# 2. Define the Advanced Dynamic Pricing Engine with Safety Guardrails
def advanced_pricing_engine(row):
    """
    Evaluates stock levels, competitor pricing, and enforces a strict cost floor 
    to protect profit margins.
    """
    cost = row['our_cost']
    comp_price = row['competitor_price']
    stock = row['stock_level']
    current_price = row['our_price']
    
    # Guardrail: Absolute minimum acceptable price (ensures at least a 15% profit margin over cost)
    min_allowable_price = round(cost * 1.15, 2)
    
    # Rule 1: Low stock scarcity -> Increase price slightly to protect inventory
    if stock < 15:
        target_price = round(comp_price * 1.02, 2)
        action = "Increase (Low Stock)"
        
    # Rule 2: Competitor is cheaper AND we have high stock -> Undercut them to drive sales
    elif comp_price < current_price and stock > 50:
        target_price = round(comp_price - 1.00, 2)
        action = "Match & Undercut"
        
    # Rule 3: Otherwise, hold our current price steady
    else:
        target_price = current_price
        action = "Hold Price"
        
    # Safety Check: Never let the target price drop below our minimum cost floor!
    if target_price < min_allowable_price:
        return min_allowable_price, f"Floor Protected (Min: {min_allowable_price})"
    
    return target_price, action

# 3. Apply our dynamic pricing function row by row across the DataFrame
advanced_results = df_pricing.apply(advanced_pricing_engine, axis=1)

# Unpack the returned tuples into clean, dedicated columns
df_pricing['optimized_price'] = [r[0] for r in advanced_results]
df_pricing['pricing_strategy'] = [r[1] for r in advanced_results]

# Calculate the projected profit per unit at the new optimized price
df_pricing['projected_profit'] = round(df_pricing['optimized_price'] - df_pricing['our_cost'], 2)

# 4. Display the final professional-grade pricing table
df_pricing
```

```python
import pandas as pd
import numpy as np

# 1. Generate a realistic, large-scale e-commerce dataset (1,000 products)
np.random.seed(42) # Ensures reproducibility
n_products = 1000

categories = ["Electronics", "Home & Living", "Apparel", "Beauty", "Sports"]

data = {
    'product_id': [f"PROD_{1000 + i}" for i in range(n_products)],
    'category': np.random.choice(categories, n_products),
    'our_cost': np.round(np.random.uniform(10.0, 300.0, n_products), 2),
    'competitor_price': np.round(np.random.uniform(15.0, 450.0, n_products), 2),
    'stock_level': np.random.randint(2, 200, n_products),
    'historical_sales_volume': np.random.randint(10, 500, n_products) # Units sold last month
}

df_catalog = pd.DataFrame(data)

# Set our baseline current price to be a markup over our cost
df_catalog['our_price'] = np.round(df_catalog['our_cost'] * np.random.uniform(1.2, 1.8, n_products), 2)

# 2. Advanced Dynamic Pricing Engine with Margin Guardrails & Scarcity Logic
def professional_pricing_engine(row, min_margin_pct=15.0):
    cost = row['our_cost']
    comp_price = row['competitor_price']
    stock = row['stock_level']
    current_price = row['our_price']
    sales_velocity = row['historical_sales_volume']
    
    # Absolute minimum price floor to guarantee profit margin
    min_allowable_price = round(cost * (1 + (min_margin_pct / 100.0)), 2)
    
    # Rule 1: Low stock scarcity (high demand, low supply) -> Increase price to protect inventory
    if stock < 15 and sales_velocity > 100:
        target_price = round(comp_price * 1.03, 2)
        action = "Increase (High Demand / Low Stock)"
        
    # Rule 2: High stock & competitor is cheaper -> Undercut competitor to clear inventory
    elif comp_price < current_price and stock > 100:
        target_price = round(comp_price - 1.00, 2)
        action = "Match & Undercut"
        
    # Rule 3: Steady state -> Maintain competitive positioning
    else:
        target_price = current_price
        action = "Hold Price"
        
    # Safety Check: Never let the price break below our margin floor
    if target_price < min_allowable_price:
        return min_allowable_price, f"Floor Protected (Min: ${min_allowable_price})"
    
    return target_price, action

# 3. Apply the engine across all 1,000 products efficiently
results = df_catalog.apply(professional_pricing_engine, axis=1)

df_catalog['optimized_price'] = [r[0] for r in results]
df_catalog['pricing_strategy'] = [r[1] for r in results]

# Calculate projected financial impact
df_catalog['projected_profit_per_unit'] = round(df_catalog['optimized_price'] - df_catalog['our_cost'], 2)
df_catalog['total_projected_revenue'] = round(df_catalog['optimized_price'] * df_catalog['historical_sales_volume'], 2)

# Preview the first few rows of our enterprise-scale data
print(df_catalog[['product_id', 'category', 'our_cost', 'our_price', 'optimized_price', 'pricing_strategy']].head(10))
# Save our 1,000-row catalog to a CSV file in your project folder
df_catalog.to_csv("ecommerce_pricing_catalog.csv", index=False)
print("Catalog successfully exported to ecommerce_pricing_catalog.csv!")
```
