```python
import pandas as pd
import numpy as np
np.random.seed(42)

data = {
    'product_id': ["PROD_101", "PROD_102", "PROD_103", "PROD_104", "PROD_105"],
    'category': ["Electronics", "Home & Living", "Electronics", "Apparel", "Home & Living"],
    'our_price': [120.50, 85.00, 210.00, 45.99, 150.00],
    'competitor_price': [130.00, 80.00, 205.00, 50.00, 140.00],
    'stock_level': [15, 120, 8, 250, 45]  
}

df = pd.DataFrame(data)

def calculate_optimal_price(row):
    if row['stock_level'] < 15:
        return round(row['competitor_price'] * 1.02,2),"Increase (Low Stock)"
    elif row['competitor_price'] < row['our_price'] and row['stock_level'] > 50:
        return round(row['competitor_price'] - 1.00,2), "Match & Undercut"
    else:

     return row['our_price'], "Hold Price"
results = df.apply(calculate_optimal_price,axis=1)
df['recommended_price'] = [r[0] for r in results]
df['pricing_action'] = [r[1] for r in results]
df
       
```
